MapReduce job using the mrjob library in Python. It goes beyond a simple word count by adding a secondary step to sort the results and filter for the top
This is formatted as code

 100 most frequent words.

```python
%%writefile word_count_3steps.py

from mrjob.job import MRJob
from mrjob.step import MRStep


class MRWordFrequency(MRJob):

    # Step 3 — Define MapReduce Steps
    def steps(self):
        return [
            MRStep(
                mapper=self.mapper_words,
                reducer=self.reducer_count
            ),
            MRStep(
                reducer=self.reducer_sort
            )
        ]


    # Step 1 — Mapper
    # Reads dataset line by line
    # Splits words
    # Emits (word,1)
    def mapper_words(self, _, line):
        for word in line.split():
            yield word.lower(), 1


    # Step 2 — Reducer
    # Groups words and counts frequency
    def reducer_count(self, word, counts):
        yield None, (sum(counts), word)


    # Step 3 — Final Output
    # Sort descending and extract top 100 words
    def reducer_sort(self, _, word_counts):
        top_100 = sorted(word_counts, reverse=True)[:100]
        for count, word in top_100:
            yield word, count


if __name__ == '__main__':
    MRWordFrequency.run()
```
